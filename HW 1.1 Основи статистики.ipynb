```python
# Завдання 1. Ми обговорили з Вами на лекції, що велика літера Сігма грецького алфавіту позначає в математиці суму. Знайдіть з використанням Python 
# Σx ,  Σx2 , та  Σ(x−1)2  для наступного набору даних: 1 3 4 Виведіть результат кожного обчислення на екран.

x = [1, 3, 4]
sum_x = sum(x)
print(f"Sum all x: {sum_x}")

sum_x_squared = sum(i**2 for i in x)
print(f"Sum all x**2: {sum_x_squared}")

sum_x_minus1_sq = sum((i - 1)**2 for i in x)
print(f"Sum (x−1) squared: {sum_x_minus1_sq}")
```

```python
# Завдання 2. Ми зібрали дані про суму покупок в гривнях 20ма випадково обраними клієнтами в нашому магазині (всього клієнтів у нас більше) за минулий місяць. 
# Вони представлені в змінній sum_purchases. Знайдіть вибіркове середнє для цих даних кожним з зазначених методів:
# 2.1. без використання жодних бібліотек, користуючись тільки Python.

sum_purchases = [316, 493, 367, 153, 224, 121, 322, 450, 196, 397, 281, 239, 484, 124, 223, 448, 432, 255, 4310, 3524]

mean_purch_py = sum(sum_purchases)/len(sum_purchases)
print(f"Mean with Python: {mean_purch_py}")
```

```python
# 2.2. використовуючи відповідний метод бібліотеки numpy.
# Виведіть результат обчислень в кожному з випадків на екран та порівняйте результати. Вони співпадають?
import numpy as np

mean_purch_np = np.mean(sum_purchases)
print(f"Mean with Numpy: {mean_purch_np}")

print(f"\nDo 2 numbers equal: {mean_purch_py == mean_purch_np}")
```

```python
# Завдання 3. Для того ж набору даних в змінній sum_purchases обчисліть медіану кожним з зазначених методів:
# 3.1. без використання жодних бібліотек, користуючись тільки Python.
def find_mediana(purchases):
    sorted_p = sorted(purchases)
    n = len(purchases)

    if n % 2 == 0:
        return (sorted_p[n // 2 - 1] + sorted_p[n // 2 ]) / 2
    else:
        return sorted_p[n // 2]
       

mediana_python = find_mediana(sum_purchases)
print(f"Median with Python: {mediana_python}")
```

```python
# 3.2. використовуючи відповідний метод бібліотеки numpy.
# Виведіть результат обчислень в кожному з випадків на екран. Порівняйте результати зі значенням вибіркового середнього. 
# Напишіть коментар, для яких задач ви б використовували вибіркове середнє, а для яких - медіану?

mediana_numpy = np.median(sum_purchases)
print(f"Median with Numpy: {mediana_numpy}")
print(f"\nDo 2 numbers median with Python and Numpy equal: {mediana_python == mediana_numpy}")
print(f"""\nDo mean and median equal: {mean_purch_np == mediana_numpy}. 
Is mean bigger then mediana: {mean_purch_np > mediana_numpy} and how much: {mean_purch_np - mediana_numpy:.2f}. 
Top 3 largest values (looking for outliers): {sorted(sum_purchases)[-3:]}. \n
Since there is a significant difference, we found outliers and should use the median to measure the typical store receipt.
However, we could track the mean over time to compare monthly performance and check if such outliers are regular or random. 
""")

# Намагаюся програмно показати, що у мене середнє значення та медіана відрізняються, отже, є аутлейери, і судити про середній чек магазину краще за медіаною.
# А середнє значення в цьому випадку можна було б використовувати, наприклад, для порівняння середнього значення щомісяця — так буде зрозуміло, чи такі аутлейери 
# взагалі є нормою, чи це випадковість у конкретному місяці, чи то сезонныість, чи результати маркетингової акції.
```

```python
# Завдання 4. Обчислість стандартне відхилення наведеної вибірки в змінній sum_purchases кожним з зазначених методів:
# 4.1. без використання жодних бібліотек, користуючись тільки Python.
el_minus_mean_py = [(el - mean_purch_py)**2 for el in sum_purchases]
std_purch_sample_py = (sum(el_minus_mean_py) / (len(sum_purchases) - 1))**0.5
print(f"Standard deviation with Python: {std_purch_sample_py:.2f}\nHere and throughout this lab, I apply Bessel's correction / Unbiased Estimator ((n - 1) instead of n), since this is a sample and a very small one: fewer than 30.")
```

```python
# 4.2. використовуючи відповідний метод бібліотеки numpy.
# Виведіть результат обчислень в кожному з випадків на екран.
std_purch_sample_np = np.std(sum_purchases, ddof = 1)
print(f"Standard deviation with Numpy: {std_purch_sample_np:.2f}")

print(f"\nDo std with Py and std with NP equal: {std_purch_sample_py == std_purch_sample_np}")
```

```python
# Завдання 5. Ми провели дослідження віку студентів-першокурсників. Виявилось, що
# 10 студентів мають вік 16 років
# 25 студентів - 17 років
# 20 студентів - 18 років
# 2 студенти - 20 років
# Обчислість середнє значення для цієї вибірки будь-яким способом - на чистому Python або з numpy - та виведіть результат на екран.

students = [16] * 10 + [17] * 25 + [18] * 20 + [20] * 2
print(f"Students age: {students}")

mean_age = np.mean(students)
print(f"Mean students age: {mean_age:.2f}")
```

```python
# Завдання 6. Крім даних про суму покупок, ми зібрали ще дані про кількість відвідувань кожного з кожного покупця, які знаходяться в змінній visits.
#  Обчисліть коефіцієнт кореляції Пірсона для суми покупок і кількості відвідувань кожним з зазначених методів:
# 6.1. без використання жодних бібліотек, користуючись тільки Python.
# і зробіть заключення, чи варто щось змінити при підрахунку кореляції, аби коефіцієнт Пірсона показував коректну оцінку лінійної залежності?
visits = [ 6, 2, 10, 3, 9, 9, 2, 3, 6, 6, 10, 2, 5, 10, 3, 5, 4, 7, 4, 2]

mean_visits = sum(visits) / len(visits)
purch_vis = [(sum_purchases[i] - mean_purch_py) * (visits[i] - mean_visits) for i in range(len(sum_purchases))]

cov_purch_vis = sum(purch_vis) / (len(sum_purchases) - 1)
print(f"Covariance with Python: {cov_purch_vis:.2f}")

visit_minus_mean = [(visit - mean_visits)**2 for visit in visits]
std_visits_sample = (sum(visit_minus_mean)/(len(visits)-1)) ** 0.5

pirr = cov_purch_vis / (std_visits_sample * std_purch_sample_py)
print(f"Pearson correlation coefficient with Python: {pirr:.2f}")
```

```python
# 6.2. використовуючи відповідний метод бібліотеки numpy.
# Виведіть результат на екран. Проаналізуйте результат.
# Виведіть додатково графік розсіювання викликавши код нижче 
pirr_np = np.corrcoef(sum_purchases, visits)[0,1]
print(f"Pearson correlation coefficient with Numpy: {pirr_np:.2f}\n")
print("According to the Cheddock Scale, a Pearson correlation coefficient ranging from 0.1 to 0.3 indicates a weak linear relationship. A negative sign indicates an inverse relationship: the higher the total purchase amount, the fewer the visits.")
```

```python
import matplotlib.pyplot as plt

plt.scatter(sum_purchases, visits)
plt.title('Scatter Plot of Purchases vs. Visits')
plt.xlabel('Sum of Purchases')
plt.ylabel('Number of Visits');

print("The scatter plot shows two outliers that affect both the plot itself and the Pearson correlation coefficient:")
```

```python
# Завдання 7. В попередньому завданні ми помітили, що викиди нам заважають отримати "правдиве" значення кореляції. Давайте вилучимо записи, які відповідають значенням 
# з викидами в сумі покупок і обчисліть коеф. кореляції. Для цього поставте threshold, який дорінює середньому значенню + 2 стандартних вдіхилля суми покупок 
# і всі значення, які більше за нього, будемо вважати викидами (є і інші правила для визначення викидів - з ним и познайомимось пізніше).
# Вилучіть значення з обох масивів і обчисліть коеф. кореляції Пірсона з допомогою numpy для кількості візитів і суми покупок без записів-викидів.
# Виведіть результат на екран, побудуйте діаграму розсіювання і проаналізуйте результат: що можемо сказати про залежність в даних?

purch_arr = np.array(sum_purchases)
visits_arr = np.array(visits)

threshold = mean_purch_np + 2 * std_purch_sample_np
print(f"Threshold: {threshold:.2f}")

mask = purch_arr <= threshold
filtered_purchases = purch_arr[mask]
filtered_visits = visits_arr[mask]

clean_corr = np.corrcoef(filtered_purchases, filtered_visits)[0, 1]
print(f"Pearson correlation without outliers: {clean_corr:.2f}")
print(f"Compare two coefficients Pearson correlation with and without outliers, is the last one bigger in absolute value: {abs(clean_corr) > abs(pirr_np)}")
print("According to the Cheddock Scale, a Pearson correlation coefficient ranging from 0.3 to 0.5 indicates a noticeable linear relationship. \nA negative sign indicates an inverse relationship: the higher the total purchase amount, the fewer the visits. Or the larger the receipt, the fewer the visits.")
```
