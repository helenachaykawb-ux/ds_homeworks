```python
# 1. Функція з параметром
def show_menu(dishes):
    print("=== МЕНЮ ===")
    for i, dish in enumerate(dishes, start=1):
        print(f"{i}. {dish}")
    print("=============")
    
menu = ["Суп грибний", "Котлета по-київськи", "Лимонад"]
show_menu(menu)
```

```python
# 2. Функція з параметром
import random

def print_square(n):
    return n**2
      
random.seed(4)
nums = [random.randint(1, 60) for i in range(5)]
print(f"My numbers for example: {nums} \n")

for num in nums:
    print(f"Квадрат числа {num} = {print_square(num)}")
```

```python
#3. Функція з поверненням значення

def multiply(a, b):
    return a * b

random.seed(4)
num1 = random.randint(1, 60) 
print(f"My first random number {num1 = }")
num2 = random.randint(1, 60)
print(f"My second random number {num2 = } \n")

print(f"And the result of multiplying numbers: {multiply(num1, num2)}")
```

```python
#4. Функція пошуку максимального значення

def find_max(numbers):
    return max(numbers)

my_list = [15, 42, 8, 23, 56, 19]
max_value = find_max(my_list)
print(f"Congratulation! We found max number from the list {my_list}: {max_value}")
```

```python
#5. Функція з умовами

def check_grade(grade):
    if grade >= 60:
        return "passed"
    return "didn't pass"

random.seed(7)
nums = [random.randint(1, 100) for i in range(10)]
print(f"My list of grades for example: {nums} \n")

for num in nums:
    print(f"The grade is {num}, so the student {check_grade(num)}")
```
