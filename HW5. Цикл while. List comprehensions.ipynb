```python
# 1. Цикл while
counter = -3

while counter <= 5:
    print(f"{counter = }")
    counter += 1
```

```python
#2. List comprehension - прості операції
numbers = [11, 4, 9, 16, 7, 5]

squares = [x**2 for x in numbers]
print(f"{squares = }")

even = [x for x in numbers if x % 2 == 0]
print(f"{even = }")

product = [x*10 for x in numbers] 
print(f"{product = }")
```

```python
#3. List comprehension - комбінована логіка
numbers = list(range(1, 11))

new_list = [x*2 if x % 2 == 0 else x - 1 for x in numbers]
print(f"{new_list = }")
```
