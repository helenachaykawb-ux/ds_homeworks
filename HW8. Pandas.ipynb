```python
# 2. Базова вибірка даних

# Колонку total_bill (загальний рахунок)
total_bill = df['total_bill']
total_bill
```

```python
# 3. Фільтрація даних

# Всі рахунки більше 20 доларів (total_bill > 20)
more_20 = df[df["total_bill"] > 20]
print(f"There are {len(more_20)} bills.\n \nFirst five bills:")
display(more_20.iloc[0:5])
```

```python
# Всі відвідувачі-жінки (sex == 'Female')
women = df[df["sex"] == "Female"]
print(f"Total female bills: {len(women)}. \n\nFirst 5 rows:")
display(women.iloc[0:5])
```

```python
# Рахунки на вечерю (time == 'Dinner')
dinner = df[df["time"] == "Dinner"]
print(f"All bills for dinner: {len(dinner)}. \n\nFirst 5 rows:")
display(dinner.iloc[0:5])
```

```python
# Рахунки на вечерю з сумою чеку більше 30 доларів
dinner_30 = df[(df["time"] == "Dinner") & (df["total_bill"] > 30)]
print(f"Dinner bills over 30 dollars: {len(dinner_30)}. \n\nFirst 5 rows:")
display(dinner_30.iloc[0:5])
```

```python
# 4. Описова статистика
# Середній розмір рахунку на всьому наборі даних
print(f"Average bill: {df["total_bill"].mean().round(2)} dollars")
```

```python
# Максимальні чайові
print(f"Max tips: {df["tip"].max().round(2)} dollars.")
```

```python
# Кількість пропущених значень у всіх колонках
print(f"Empty rows: \n\n{df.isna().sum()}")
```

```python
# Скільки в наборі даних рахунківи і за вечері за обід
df.time.unique() #я не знаю, як вони написали "вечері", тому перевіряю
```

```python
bils_lun = df[df["time"] == "Lunch"]
bils_din = df[df["time"] == "Dinner"]
print(f"Bills in dinners {len(bils_din)} and lunches {len(bils_lun)} ")
```

```python
# Або просто порахувати з .value_counts():
print(f"Total amount of bills by {df["time"].value_counts()}")
```

```python
# 5. Статистика із групуванням
# Середній рахунок за статтю (groupby sex)
mean_bysex = df.groupby("sex")["total_bill"].mean()
print(f"Mean bill by {mean_bysex}")
```

```python
# Середні чайові за днем тижня (groupby day)
mean_tips = df.groupby("day")["tip"].mean().round(2)
print(f"Average tips by: {mean_tips}")
```

```python
# Скільки відвідувачів у кожен день
day_people = df.groupby("day")["size"].sum()
print(f"Total visitors per {day_people}")
```

```python
# 6. Створення нової колонки
df["tip_percentage"] = df["tip"] * 100 / df["total_bill"]
df.head(10)
```

```python
# 7. Сортування
# За розміром рахунку (зростання)
df_sorted_bill = df.sort_values("total_bill")
df_sorted_bill.head()
```

```python
# За чайовими (спадання)
df_sorted_tips = df.sort_values("tip", ascending=False)
df_sorted_tips.head()
```

```python
# Додатково виведіть топ-5 найбільших чайових використовуючи метод nlargest.
df_top_5_tips = df.nlargest(5, "tip")
df_top_5_tips
```
