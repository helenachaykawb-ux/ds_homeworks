```python
#1. Цикл for - базовий

for i in range(1, 8):
    print(i)
```

```python
#2. Цикл for по списку

names = ["Олег", "Анна", "Іван", "Марія"]

for name in names:
    print(f"Привіт, {name}!")
```

```python
#3. Цикл з enumerate

grades = [85, 92, 78, 95, 88]

for num, grade in enumerate(grades, start=1):
    print(f"Студент {num}: {grade} балів")
```

```python
#4. Сума парних чисел

sum_even = 0

for i in range(1, 21):
    if i % 2 == 0: 
        sum_even += i

print(sum_even)
```
