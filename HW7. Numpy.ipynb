```python
import numpy as np 

# 1. Створення масивів

arr = np.array([5, 15, 25, 35, 45])
print(f"{arr=}")

zeros = np.zeros(7, )
print(f"{zeros=}")

ones = np.ones(6, )
print(f"{ones=}")

arr2 = np.arange(10, 20)
print(f"{arr2=}")
```

```python
# 2. Базові операції

arr = np.array([12, 24, 36, 48, 60])

print(f"{arr=}\n")

mult = arr * 2
print(f"{mult=}")

add_10 = arr + 10
print(f"{add_10=}")

sqr = arr ** 2
print(f"{sqr=}")

sqrt = np.sqrt(arr)
print(f"{sqrt=}")
```

```python
# 3. Статистичні функції

data = np.array([18, 27, 33, 41, 52, 68, 79])

print(f"{data=}\n")

print(f"{np.mean(data)=}")
print(f"{np.min(data)=}")
print(f"{np.max(data)=}")
print(f"{np.sum(data)=}")
```

```python
# 4. Індексація

arr = np.array( [15, 22, 35, 44, 58, 67])

print(f"{arr=}\n")

print(f"{arr[0]=}")
print(f"{arr[-1]=}")
print(f"{arr[1:4]=}") # "Елементи з індексами 1 до 3" - вважаю, що включно індекс 3!
print(f"{arr[arr > 40]=}")
```

```python
# 5. Двовимірні масиви

matrix = np.array([[2, 4, 6],
                   [8, 10, 12],
                   [14, 16, 18]])

print(f"{matrix=}\n")

print(f"{matrix.shape=}")
print(f"{matrix[0, ]=}")
print(f"{matrix[:, 1]=}")
print(f"{np.sum(matrix)=}")
print(f"{np.mean(matrix)=}")
```
