```python
import numpy as np 

# 1. Створення масивів

arr = np.array([5, 15, 25, 35, 45])
display(arr)

zeros = np.zeros(7, )
display(zeros)

ones = np.ones(6, )
display(ones)

arr2 = np.arange(10, 20)
display(arr2)
```

```python
# 2. Базові операції

arr = np.array([12, 24, 36, 48, 60])

m = arr * 2
display(m)

s = arr + 10
display(s)

sqr = arr ** 2
display(sqr)

sqrt = np.sqrt(arr)
display(sqrt)
```

```python
# 3. Статистичні функції

data = np.array([18, 27, 33, 41, 52, 68, 79])

display(np.mean(data))
display(np.min(data))
display(np.max(data))
display(np.sum(data))
```

```python
# 4. Індексація

arr = np.array( [15, 22, 35, 44, 58, 67])

display(arr[0])
display(arr[-1])
display(arr[1:4]) # "Елементи з індексами 1 до 3" - вважаю, що включно індекс 3!
display(arr[arr > 40])
```

```python
# 5. Двовимірні масиви

matrix = np.array([[2, 4, 6],
                   [8, 10, 12],
                   [14, 16, 18]])

display(matrix.shape)
display(matrix[0, ])
display(matrix[:, 1])
display(np.sum(matrix))
display(np.mean(matrix))
```
