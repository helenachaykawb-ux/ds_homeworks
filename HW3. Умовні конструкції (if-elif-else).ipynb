```python
grade = 72

if grade >= 90:
    print("Відмінно")
elif grade >= 75: #elif 75 <= grade < 90: оце прекрати писать
    print("Добре")
elif grade >= 60: #оце прекрати писать двойные проверки где не надо
    print("Задовільно")
else:
    print("Незадовільно")
```
